Sekarang Hasil Dot Product Dikalikan dengan $q_1$

Diketahui:
$$q_1 =\begin{bmatrix}\dfrac{2\sqrt{5}}{5} \\\dfrac{\sqrt{5}}{5}\end{bmatrix}$$
dan hasil dot product:
$$q_1 \cdot a_2 = \dfrac{4\sqrt{5}}{5}$$
Sekarang kalikan:
$(q_1 \cdot a_2)q_1$
$$=\frac{4\sqrt{5}}{5}\begin{bmatrix}\dfrac{2\sqrt{5}}{5} \\\dfrac{\sqrt{5}}{5}\end{bmatrix}$$
Perhitungan:
$$=\begin{bmatrix}\dfrac{4\sqrt{5}}{5} \times \dfrac{2\sqrt{5}}{5} \\\dfrac{4\sqrt{5}}{5} \times \dfrac{\sqrt{5}}{5}\end{bmatrix}$$
$$=\begin{bmatrix}\dfrac{8}{5} \\\dfrac{4}{5}\end{bmatrix}$$

```python
from sympy import Matrix, sqrt

# Matriks A
A = Matrix([
    [2, 1],
    [1, 2]
])

# Ambil kolom
a1 = A[:, 0]
a2 = A[:, 1]

# q1
panjang = sqrt(5)
q1 = a1 / panjang

# dot product
dot = q1.dot(a2)

# hasil dot * q1
hasil = dot * q1

print("Hasil =")
print(hasil)
```

hasil dot product yang dikalikan $q_1$ di kurangi dengan $a_2$

Diketahui:
$$a_2 =\begin{bmatrix}1 \\2\end{bmatrix}$$
dan hasil sebelumnya:
$$(q_1 \cdot a_2)q_1 =\begin{bmatrix}\dfrac{8}{5} \\\dfrac{4}{5}\end{bmatrix}$$
Sekarang hitung:
$$u_2 = a_2 - (q_1 \cdot a_2)q_1$$

$$u_2 =\begin{bmatrix}1 \\2\end{bmatrix}-\begin{bmatrix}\dfrac{8}{5} \\\dfrac{4}{5}\end{bmatrix}$$
Perhitungan:

$$=\begin{bmatrix}1 - \dfrac{8}{5} \\2 - \dfrac{4}{5}\end{bmatrix}$$

$$=\begin{bmatrix}\dfrac{5}{5} - \dfrac{8}{5} \\\dfrac{10}{5} - \dfrac{4}{5}\end{bmatrix}$$

$$=\begin{bmatrix}-\dfrac{3}{5} \\\dfrac{6}{5}\end{bmatrix}$$

```python
from sympy import Matrix, sqrt

# Matriks A
A = Matrix([
    [2, 1],
    [1, 2]
])

# Ambil kolom
a1 = A[:, 0]
a2 = A[:, 1]

# q1
q1 = a1 / sqrt(5)

# Proyeksi
dot = q1.dot(a2)
proyeksi = dot * q1

# u2
u2 = a2 - proyeksi

print("u2 =")
print(u2)
```

Sekarang Kita akan Mencari Matrix R

Karena tadi kita sudah mendapatkan:

$$
q_1 =
\begin{bmatrix}
\dfrac{2\sqrt{5}}{5} \\
\dfrac{\sqrt{5}}{5}
\end{bmatrix}
$$

dan

$$
q_2 =
\begin{bmatrix}
-\dfrac{\sqrt{5}}{5} \\
\dfrac{2\sqrt{5}}{5}
\end{bmatrix}
$$

Maka matriks Q adalah:

$$
Q =
\begin{bmatrix}
\dfrac{2\sqrt{5}}{5} & -\dfrac{\sqrt{5}}{5} \\
\dfrac{\sqrt{5}}{5} & \dfrac{2\sqrt{5}}{5}
\end{bmatrix}
$$

Untuk mencari matriks R:

$$
R = Q^T A
$$

Dengan:

$$
A =
\begin{bmatrix}
2 & 1 \\
1 & 2
\end{bmatrix}
$$

dan

$$
Q^T =
\begin{bmatrix}
\dfrac{2\sqrt{5}}{5} & \dfrac{\sqrt{5}}{5} \\
-\dfrac{\sqrt{5}}{5} & \dfrac{2\sqrt{5}}{5}
\end{bmatrix}
$$

Hasil perkalian:

$$
R =
\begin{bmatrix}
\sqrt{5} & \dfrac{4\sqrt{5}}{5} \\
0 & \dfrac{3\sqrt{5}}{5}
\end{bmatrix}
$$

```python
from sympy import Matrix, sqrt

# Matriks A
A = Matrix([
    [2, 1],
    [1, 2]
])

# q1 dan q2
q1 = Matrix([
    [2*sqrt(5)/5],
    [sqrt(5)/5]
])

q2 = Matrix([
    [-sqrt(5)/5],
    [2*sqrt(5)/5]
])

# Matriks Q
Q = Matrix.hstack(q1, q2)

print("Q =")
print(Q)

# Matriks R
R = Q.T * A

print("\nR =")
print(R)
```

Cari $A_{k+1} = R_kQ_k$

Diketahui:

$$R = \begin{bmatrix} \sqrt{5} & \frac{4\sqrt{5}}{5} \\ 0 & \frac{3\sqrt{5}}{5} \end{bmatrix}$$

dan

$$Q = \begin{bmatrix} \frac{2\sqrt{5}}{5} & -\frac{\sqrt{5}}{5} \\ \frac{\sqrt{5}}{5} & \frac{2\sqrt{5}}{5} \end{bmatrix}$$

Sekarang hitung:

$$RQ = \begin{bmatrix} \sqrt{5} & \frac{4\sqrt{5}}{5} \\ 0 & \frac{3\sqrt{5}}{5} \end{bmatrix} \begin{bmatrix} \frac{2\sqrt{5}}{5} & -\frac{\sqrt{5}}{5} \\ \frac{\sqrt{5}}{5} & \frac{2\sqrt{5}}{5} \end{bmatrix}$$

Baris 1 Kolom 1

$$(\sqrt{5} \times \frac{2\sqrt{5}}{5}) + (\frac{4\sqrt{5}}{5} \times \frac{\sqrt{5}}{5})$$ $$= 2 + \frac{4}{5}$$ $$= \frac{14}{5}$$

Baris 1 Kolom 2

$$(\sqrt{5} \times -\frac{\sqrt{5}}{5}) + (\frac{4\sqrt{5}}{5} \times \frac{2\sqrt{5}}{5}) $$ $$= -1 + \frac{8}{5}$$ $$= \frac{3}{5}$$

Baris 2 Kolom 1

$$(0 \times \frac{2\sqrt{5}}{5}) + (\frac{3\sqrt{5}}{5} \times \frac{\sqrt{5}}{5})$$ $$= 0 + \frac{3}{5}$$ $$= \frac{3}{5}$$

Baris 2 Kolom 2

$$(0 \times -\frac{\sqrt{5}}{5}) + (\frac{3\sqrt{5}}{5} \times \frac{2\sqrt{5}}{5})$$ $$= 0 + \frac{6}{5}$$ $$= \frac{6}{5}$$

Hasil Akhir

$$RQ = \begin{bmatrix} \frac{14}{5} & \frac{3}{5} \\ \frac{3}{5} & \frac{6}{5} \end{bmatrix}$$

### Penjelasan Proses QR Iteration

### Metode pada gambar disebut QR Iteration.

### Tujuannya untuk mendekati nilai eigen matriks.

###Langkah-langkahnya:
### 1. Mulai dari matriks awal
$$A_0 = A$$
Contoh:
$$A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}$$

### 2. Faktorkan matriks menjadi $Q$ dan $R$
$$A_k = Q_k R_k$$
* $Q$ = matriks ortogonal
* $R$ = matriks segitiga atas

Pada kode:
* $q_1$ dan $q_2$ dicari menggunakan Gram-Schmidt
* Lalu digabung menjadi matriks $Q$:
$$Q = \begin{bmatrix} q_1 & q_2 \end{bmatrix}$$
Kemudian:
$$R = Q^T A$$
Jadi:
$$A_k = Q_k R_k$$

### 3. Balik urutan perkalian
Setelah mendapatkan $Q$ dan $R$:
$$A_{k+1} = R_k Q_k$$
$$A_{k+1} = R_k Q_k$$
Ini inti dari QR Iteration.

### 4. Ulangi proses
Matriks baru:
$$A_k$$
digunakan sebagai iterasi berikutnya. Semakin banyak iterasi:
* elemen di bawah diagonal akan mendekati 0
* diagonal matriks akan mendekati nilai eigen

### Hasil yang Terjadi
Iterasi akan menghasilkan matriks yang mendekati:
$$\begin{bmatrix} 3 & 0 \\ 0 & 1 \end{bmatrix}$$
Di mana angka:
* $3$
* $1$
adalah nilai eigen matriks awal.

```python
from sympy import Matrix, sqrt, N, pprint

# =========================
# Matriks awal
# =========================
A = Matrix([
    [2, 1],
    [1, 2]
])

print("A0 =")
pprint(A)

# =========================
# QR ITERATION
# =========================
for k in range(1, 11):

    # Ambil kolom
    a1 = A[:, 0]
    a2 = A[:, 1]

    # q1
    q1 = a1 / sqrt(a1.dot(a1))

    # proyeksi
    proj = q1.dot(a2) * q1

    # v2
    v2 = a2 - proj

    # q2
    q2 = v2 / sqrt(v2.dot(v2))

    # Matriks Q
    Q = Matrix.hstack(q1, q2)

    # Matriks R
    R = Q.T * A

    # Matriks baru
    A = R * Q

    print(f"\nA{k} =")
    pprint(N(A, 5))
```
